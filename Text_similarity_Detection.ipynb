```python
import torch
import gradio as gr
from transformers import AutoTokenizer, AutoModel
from sklearn.metrics.pairwise import cosine_similarity
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")
model.eval()
```

```python
def text_similarity(text1, text2):
    inputs1 = tokenizer(text1, return_tensors="pt", padding=True, truncation=True)
    inputs2 = tokenizer(text2, return_tensors="pt", padding=True, truncation=True)

    with torch.no_grad():
        output1 = model(**inputs1).last_hidden_state.mean(dim=1)
        output2 = model(**inputs2).last_hidden_state.mean(dim=1)

    score = cosine_similarity(output1.numpy(), output2.numpy())[0][0]

    return round(float(score), 4)
```

```python
app = gr.Interface(
    fn=text_similarity,
    inputs=[
        gr.Textbox(label="Text 1", placeholder="Enter first sentence"),
        gr.Textbox(label="Text 2", placeholder="Enter second sentence")
    ],
    outputs=gr.Number(label="Similarity Score"),
    title="Text Similarity Detection",
    description="This app uses Hugging Face Transformers to measure semantic similarity between two texts."
)
```

```python
app.launch(share=True)
```
