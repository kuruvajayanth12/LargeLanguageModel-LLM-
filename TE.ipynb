```python
import nltk
import random
from nltk.tokenize import word_tokenize

# Download required resources (run once)
nltk.download('punkt')

# Take input from user
text = input("Enter the text: ")

# Word-based tokenization
words = word_tokenize(text)

print("\nTokenized Words:")
print(words)

# ----------------------------
# Assign random numbers to words
# ----------------------------

word_to_random = {}
numerical_tokens = []

for word in words:
    if word not in word_to_random:
        word_to_random[word] = random.randint(1, 1000)
    numerical_tokens.append(word_to_random[word])

print("\nWord → Random Number Mapping:")
print(word_to_random)

print("\nNumerical Representation:")
print(numerical_tokens)
```

```python
import numpy as np

# ----------------------------
# Map words to their 3-D embeddings
# ----------------------------

embedding_dim = 3

number_to_embedding = {}
word_embeddings = []

for word, num in zip(words, numerical_tokens):
    if num not in number_to_embedding:
        number_to_embedding[num] = np.random.rand(embedding_dim)
    word_embeddings.append((word, number_to_embedding[num]))

print("\nWord → 3-D Embedding:\n")
for word, vector in word_embeddings:
    print(f"{word} : {vector}")
```
