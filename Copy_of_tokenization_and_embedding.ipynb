```python
from nltk.tokenize import word_tokenize
import numpy as np

text = input("Enter the text:")
tokens = word_tokenize(text)
print("Tokens:", tokens)

embedding_dim = 3
embeddings = {}

for token in tokens:
    embeddings[token] = np.random.rand(embedding_dim)

print("\nToken Embeddings:")
for token, vector in embeddings.items():
    print(token, ":", vector)
```
